```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from google.colab import files

uploaded = files.upload()
dataset = pd.read_csv(next(iter(uploaded)))

display(dataset)
print(dataset.columns)
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)

indep = dataset[['R&D Spend', 'Administration', 'Marketing Spend',
                 'State_Florida', 'State_New York']]

dep = dataset[['Profit']]

display(indep)
display(dep)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    indep, dep, test_size=1/3, random_state=0
)

display(X_train)
display(y_train)
display(X_test)
display(y_test)

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, y_train)

weight = regressor.coef_
print("Weight of the model={}".format(weight))

bais = regressor.intercept_
print("Intercept of the model={}".format(bais))
```

```python
from sklearn.metrics import r2_score

y_pred = regressor.predict(X_test)

r_score = r2_score(y_test, y_pred)

print("R2 Score:", r_score)
```

```python
import pickle

filename = "finalized_model_Mul_linear.sav"

pickle.dump(regressor, open(filename, 'wb'))

print("Model saved successfully!")
```

```python
loaded_model = pickle.load(
    open("finalized_model_Mul_linear.sav", 'rb')
)

result = loaded_model.predict([[1234, 345, 4565, 1, 0]])

print("Predicted Profit:", result)
```
